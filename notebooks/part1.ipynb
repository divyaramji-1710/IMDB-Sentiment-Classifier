```python
# ============================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ============================================================

# Pandas is used for:
# - Loading the dataset
# - Storing data in a DataFrame
# - Inspecting and manipulating tabular data
import pandas as pd

# NumPy is used for:
# - Numerical operations
# - Statistical calculations
# - Working with arrays and numerical data
import numpy as np

print("Libraries imported successfully!")
```

```python
# ============================================================
# STEP 2: LOAD THE IMDB DATASET
# ============================================================

from pathlib import Path

# Define the expected dataset locations.
# When the notebook is eventually inside the 'notebooks'
# folder, the second path will be used.
possible_paths = [
    Path("data/IMDB Dataset.csv"),
    Path("../data/IMDB Dataset.csv")
]

# Find the first path that actually exists.
dataset_path = None

for path in possible_paths:
    if path.exists():
        dataset_path = path
        break

# Stop with a clear error message if the dataset cannot be found.
if dataset_path is None:
    raise FileNotFoundError(
        "IMDB Dataset.csv was not found. "
        "Please make sure it is inside the project's 'data' folder."
    )

# Load the CSV file into a Pandas DataFrame.
df = pd.read_csv(dataset_path)

print("Dataset loaded successfully!")
print("Dataset location:", dataset_path)
```

```python
# ============================================================
# VERIFY THE DATASET
# ============================================================

# Display the first 5 rows of the dataset.
# This allows us to visually verify that the data was loaded
# correctly.
df.head()
```

```python
# ============================================================
# CHECK DATASET DIMENSIONS
# ============================================================

# df.shape returns:
# (number_of_rows, number_of_columns)
print("Dataset shape:", df.shape)
# Display the number of rows and columns in the dataset.
rows, columns = df.shape

print("Number of rows    :", rows)
print("Number of columns :", columns)
```

```python
# ============================================================
# CHECK COLUMN NAMES
# ============================================================

# Display the names of all columns in the dataset.
print("Column names:")
print(df.columns.tolist())
```

```python
# ============================================================
# COUNT THE NUMBER OF REVIEWS IN EACH SENTIMENT CLASS
# ============================================================

# value_counts() counts how many times each sentiment occurs.
label_counts = df["sentiment"].value_counts()

print("Sentiment label counts:")
print(label_counts)
```

```python
# ============================================================
# CALCULATE SENTIMENT LABEL PERCENTAGES
# ============================================================

# Calculate the percentage of the dataset belonging to each
# sentiment class.
label_percentages = df["sentiment"].value_counts(normalize=True) * 100

print("Sentiment label percentages:")
print(label_percentages.round(2))
```

```python
# ============================================================
# CALCULATE THE MAJORITY-CLASS BASELINE
# ============================================================

# Find the most common sentiment label.
majority_class = label_counts.idxmax()

# Find how many samples belong to that class.
majority_count = label_counts.max()

# Calculate the accuracy obtained by always predicting
# the majority class.
majority_baseline = (majority_count / len(df)) * 100

print("Majority class       :", majority_class)
print("Majority class count :", majority_count)
print("Majority-class baseline:", round(majority_baseline, 2), "%")
```

```python
# ============================================================
# CREATE THE WORD_COUNT COLUMN
# ============================================================

# Count the number of whitespace-separated words in each review.
#
# split() separates the review into individual words/tokens.
# len() then counts how many tokens were produced.
df["word_count"] = df["review"].str.split().str.len()

# Display the first few rows to verify the new column.
df[["review", "sentiment", "word_count"]].head()
```

```python
# ============================================================
# MEAN WORD COUNT
# ============================================================

mean_word_count = df["word_count"].mean()

print("Mean word count:", round(mean_word_count, 2))
```

```python
# ============================================================
# MEDIAN WORD COUNT
# ============================================================

median_word_count = df["word_count"].median()

print("Median word count:", median_word_count)
```

```python
# ============================================================
# CREATE WORD-COUNT BANDS
# ============================================================

# Define ranges of word counts.
# Each band contains reviews whose word counts fall within
# that range.
bins = [0, 20, 40, 60, 80, 100, 150, 200, 300, 500, np.inf]

labels = [
    "0-19",
    "20-39",
    "40-59",
    "60-79",
    "80-99",
    "100-149",
    "150-199",
    "200-299",
    "300-499",
    "500+"
]

# Assign each review to a word-count band.
df["word_count_band"] = pd.cut(
    df["word_count"],
    bins=bins,
    labels=labels,
    right=False
)

# Count how many reviews occur in each band.
mode_band_counts = df["word_count_band"].value_counts().sort_index()

print("Word-count band frequencies:")
print(mode_band_counts)
```

```python
# ============================================================
# IDENTIFY THE MODE BAND
# ============================================================

# Find the band containing the largest number of reviews.
mode_band = mode_band_counts.idxmax()

# Find the number of reviews in that band.
mode_band_frequency = mode_band_counts.max()

print("Mode band:", mode_band)
print("Number of reviews in mode band:", mode_band_frequency)
```

```python
# ============================================================
# VARIANCE AND STANDARD DEVIATION OF WORD COUNT
# ============================================================

# Calculate the sample variance.
word_count_variance = df["word_count"].var()

# Calculate the sample standard deviation.
word_count_std = df["word_count"].std()

print("Word-count variance         :", round(word_count_variance, 2))
print("Word-count standard deviation:", round(word_count_std, 2))
```

```python
# ============================================================
# SUMMARY OF WORD-COUNT STATISTICS
# ============================================================

print("WORD-COUNT STATISTICS")
print("=" * 40)

print(f"Mean                  : {mean_word_count:.2f}")
print(f"Median                : {median_word_count:.2f}")
print(f"Mode band             : {mode_band}")
print(f"Mode-band frequency   : {mode_band_frequency}")
print(f"Variance              : {word_count_variance:.2f}")
print(f"Standard deviation    : {word_count_std:.2f}")
```

```python
# ============================================================
# COMPARE MEAN AND MEDIAN
# ============================================================

difference = mean_word_count - median_word_count

print("Mean word count   :", round(mean_word_count, 2))
print("Median word count :", round(median_word_count, 2))
print("Difference        :", round(difference, 2))
```

```python
# ============================================================
# DETERMINE SKEWNESS FROM MEAN-MEDIAN RELATIONSHIP
# ============================================================

if mean_word_count > median_word_count:
    skew_direction = "right-skewed (positively skewed)"
elif mean_word_count < median_word_count:
    skew_direction = "left-skewed (negatively skewed)"
else:
    skew_direction = "approximately symmetric"

print("Distribution:", skew_direction)
```

```python
# ============================================================
# CALCULATE STATISTICAL SKEWNESS
# ============================================================

word_count_skewness = df["word_count"].skew()

print("Word-count skewness:", round(word_count_skewness, 4))
```

```python
# ============================================================
# STEP 4: MEAN WORD COUNT BY SENTIMENT
# ============================================================

# Calculate the average word count separately for
# positive and negative reviews.
mean_word_count_by_sentiment = df.groupby("sentiment")["word_count"].mean()

print("Mean word count by sentiment:")
print(mean_word_count_by_sentiment.round(2))

# Calculate the difference between positive and negative means.
positive_mean = mean_word_count_by_sentiment["positive"]
negative_mean = mean_word_count_by_sentiment["negative"]

difference = positive_mean - negative_mean

print("\nPositive mean word count :", round(positive_mean, 2))
print("Negative mean word count :", round(negative_mean, 2))
print("Difference               :", round(difference, 2))
```

```python
# ============================================================
# STEP 5: CHECK FOR DUPLICATE ROWS
# ============================================================

# Count the number of duplicate rows in the dataset.
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# ============================================================
# FINAL INTERPRETATION OF WORD-COUNT DISTRIBUTION
# ============================================================

print("INTERPRETATION")
print("=" * 60)

if word_count_skewness > 0:
    print(
        "The word-count distribution is positively (right) skewed."
    )
    print(
        "The mean is higher than the median because a relatively "
        "small number of very long reviews pull the mean upward."
    )
    print(
        "Therefore, the median is a more representative statistic "
        "for describing the typical review length."
    )

elif word_count_skewness < 0:
    print(
        "The word-count distribution is negatively (left) skewed."
    )
    print(
        "The mean is lower than the median because shorter reviews "
        "pull the mean downward."
    )
    print(
        "Therefore, the median is a more representative statistic "
        "for describing the typical review length."
    )

else:
    print(
        "The word-count distribution is approximately symmetric."
    )
    print(
        "The mean and median are therefore both reasonable measures "
        "of the typical review length."
    )
```
